```python

def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


def get_neighbors(state):

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = [
        ("RIGHT", 0, 1),
        ("DOWN", 1, 0),
        ("LEFT", 0, -1),
        ("UP", -1, 0)
    ]

    for move, dr, dc in moves:

        nr = row + dr
        nc = col + dc

        if 0 <= nr < 3 and 0 <= nc < 3:

            new_zero = nr * 3 + nc
            new_state = list(state)

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            yield move, tuple(new_state)


def dfs(state, goal, limit, visited, path):

    if state == goal:
        return True


    if limit == 0:
        return False

    visited.add(state)

    for move, next_state in get_neighbors(state):

        if next_state not in visited:

            path.append((move, next_state))

            if dfs(next_state, goal, limit - 1,
                   visited, path):
                return True
            path.pop()

    visited.remove(state)

    return False



print("1BM24CS106")
start = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

goal = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

LIMIT = 5

path = [("START", start)]

if dfs(start, goal, LIMIT, set(), path):

    for move, state in path:

        if move == "START":
            print("START")
        else:
            print("↓", move)

        print_state(state)

    print("GOAL REACHED")
```
