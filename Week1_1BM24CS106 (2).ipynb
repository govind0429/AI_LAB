```python
import random

def print_board(board):
    """Prints the Tic-Tac-Toe board."""
    for row in board:
        print(" | ".join(row))
        print("-----------")

def check_win(board, player):
    """Checks if the given player has won."""

    for row in board:
        if all([s == player for s in row]):
            return True

    for col in range(3):
        if all([board[row][col] == player for row in range(3)]):
            return True

    if all([board[i][i] == player for i in range(3)]) or \
       all([board[i][2-i] == player for i in range(3)]):
        return True
    return False

def check_draw(board):
    """Checks if the game is a draw."""
    for row in board:
        for spot in row:
            if spot == ' ':
                return False
    return True

def get_empty_cells(board):
    """Returns a list of all empty cells (row, col) on the board."""
    empty_cells = []
    for r in range(3):
        for c in range(3):
            if board[r][c] == ' ':
                empty_cells.append((r, c))
    return empty_cells

def computer_move(board):
    """Determines the computer's move using a simple AI strategy."""

    for r, c in get_empty_cells(board):
        temp_board = [row[:] for row in board]
        temp_board[r][c] = 'O'
        if check_win(temp_board, 'O'):
            return (r, c)


    for r, c in get_empty_cells(board):
        temp_board = [row[:] for row in board]
        temp_board[r][c] = 'X'
        if check_win(temp_board, 'X'):
            return (r, c)


    if board[1][1] == ' ':
        return (1, 1)


    corners = [(0, 0), (0, 2), (2, 0), (2, 2)]
    available_corners = [c for c in corners if board[c[0]][c[1]] == ' ']
    if available_corners:
        return random.choice(available_corners)


    sides = [(0, 1), (1, 0), (1, 2), (2, 1)]
    available_sides = [s for s in sides if board[s[0]][s[1]] == ' ']
    if available_sides:
        return random.choice(available_sides)

    return random.choice(get_empty_cells(board))

def play_game():

    """Plays a game of Tic-Tac-Toe."""
    board = [[' ' for _ in range(3)] for _ in range(3)]
    current_player = 'X'
    print("Govind   1BM24CS106 ")
    print("Welcome to Tic-Tac-Toe!")
    print("Player 1: X (Human)")
    print("Player 2: O (Computer)")
    print_board(board)

    while True:
        if current_player == 'X':

            while True:
                try:
                    row = int(input("Human (X), enter row (0-2): "))
                    col = int(input("Human (X), enter column (0-2): "))
                    if 0 <= row <= 2 and 0 <= col <= 2 and board[row][col] == ' ':
                        board[row][col] = 'X'
                        break
                    else:
                        print("Invalid move. Try again.")
                except ValueError:
                    print("Invalid input. Please enter a number.")
        else:

            print("Computer (O) is making a move...")
            r, c = computer_move(board)
            board[r][c] = 'O'
            print(f"Computer chose row {r}, column {c}")

        print_board(board)

        if check_win(board, current_player):
            print(f"Player {current_player} wins!")
            break
        elif check_draw(board):
            print("It's a draw!")
            break

        current_player = 'O' if current_player == 'X' else 'X'
```

```python
play_game()
```
